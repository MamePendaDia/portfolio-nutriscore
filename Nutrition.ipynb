```python
import pandas as pd

colonnes = [
    "code", "product_name", "brands", "categories", "countries",
    "quantity", "nutriscore_grade", "nutriscore_score", "nova_group",
    "environmental_score_grade", "pnns_groups_1", "pnns_groups_2",
    "energy-kcal_100g", "fat_100g", "saturated-fat_100g", "sugars_100g",
    "salt_100g", "proteins_100g", "fiber_100g",
    "additives_n", "additives_tags", "ingredients_text"
]

chemin = r"C:\Users\user\OneDrive\Documents\Nutrition\fr.openfoodfacts.org.products.csv"

chunks_filtres = []
taille_chunk = 100_000

for i, chunk in enumerate(pd.read_csv(
        chemin,
        sep="\t",
        usecols=colonnes,
        chunksize=taille_chunk,
        low_memory=False,
        on_bad_lines="skip"
    )):
    chunk_fr = chunk[chunk["countries"].astype(str).str.contains("France", case=False, na=False)]
    chunks_filtres.append(chunk_fr)
    print(f"Chunk {i+1} traité — {len(chunk_fr)} produits France retenus")

df = pd.concat(chunks_filtres, ignore_index=True)
print(f"\nTotal produits France : {len(df)}")
df.head()
```

```python
# les produits qui ont au minimum un nom ET des valeurs nutritionnelles
df_propre = df.dropna(subset=["product_name", "energy-kcal_100g"])

#Suppression des doublons éventuels sur le code produit
df_propre = df_propre.drop_duplicates(subset="code")

print(f"Avant nettoyage : {len(df)} produits")
print(f"Après nettoyage : {len(df_propre)} produits")
df_propre.head()
```

```python
chemin_sortie = r"C:\Users\user\OneDrive\Documents\Nutrition\produits_france_propre.csv"
df_propre.to_csv(chemin_sortie, index=False)
print("Fichier sauvegardé :", chemin_sortie)
```

```python
df_propre["quantity"] = df_propre["quantity"].astype(str).str[:255]
df_propre.loc[df_propre["quantity"] == "nan", "quantity"] = None

from sqlalchemy import create_engine

# Connexion à la base MySQL
utilisateur = "root"
mot_de_passe = "772728917"
hote = "localhost"
port = "3306"
base = "nutrition"

engine = create_engine(f"mysql+pymysql://{utilisateur}:{mot_de_passe}@{hote}:{port}/{base}")

# --- Table produits ---
df_produits = df_propre[[
    "code", "product_name", "quantity", "brands", "categories",
    "pnns_groups_1", "pnns_groups_2", "nutriscore_grade",
    "nutriscore_score", "nova_group", "environmental_score_grade"
]].rename(columns={
    "pnns_groups_1": "pnns_groupe1",
    "pnns_groups_2": "pnns_groupe2"
})
df_produits.to_sql("produits", engine, if_exists="append", index=False, chunksize=5000)
print("Table produits importée :", len(df_produits))

# --- Table valeurs_nutritionnelles ---
df_valeurs = df_propre[[
    "code", "energy-kcal_100g", "fat_100g", "saturated-fat_100g",
    "sugars_100g", "fiber_100g", "proteins_100g", "salt_100g"
]].rename(columns={
    "energy-kcal_100g": "energie_kcal",
    "fat_100g": "matieres_grasses",
    "saturated-fat_100g": "acides_gras_satures",
    "sugars_100g": "sucres",
    "fiber_100g": "fibres",
    "proteins_100g": "proteines",
    "salt_100g": "sel"
})
df_valeurs.to_sql("valeurs_nutritionnelles", engine, if_exists="append", index=False, chunksize=5000)
print("Table valeurs_nutritionnelles importée :", len(df_valeurs))
df_propre["quantity"] = df_propre["quantity"].astype(str).str[:255]
df_propre.loc[df_propre["quantity"] == "nan", "quantity"] = None
```

```python
df_propre = df_propre[df_propre["code"].astype(str).str.len() <= 30]
print("Lignes restantes après filtrage code :", len(df_propre))
```

```python
#les lignes qui ont des additifs renseignés
df_additifs_brut = df_propre[["code", "additives_tags"]].dropna(subset=["additives_tags"])

# Une ligne par additif au lieu d'une ligne par produit
df_additifs_brut["additives_tags"] = df_additifs_brut["additives_tags"].str.split(",")
df_additifs = df_additifs_brut.explode("additives_tags")

# Nettoyage : le préfixe "en:" et les espaces superflus
df_additifs["additives_tags"] = df_additifs["additives_tags"].str.replace("en:", "", regex=False).str.strip()
df_additifs = df_additifs.rename(columns={"additives_tags": "additif"})

# les lignes vides éventuelles
df_additifs = df_additifs[df_additifs["additif"] != ""]

print(f"Nombre total de lignes additifs : {len(df_additifs)}")
df_additifs.head(10)
```

```python
# Sécurité : tronquer les valeurs trop longues pour la colonne (VARCHAR(150))
df_additifs["additif"] = df_additifs["additif"].astype(str).str[:150]

df_additifs[["code", "additif"]].to_sql(
    "additifs", engine, if_exists="append", index=False, chunksize=5000
)
print("Table additifs importée :", len(df_additifs))
```
